```python
from PIL import Image, ImageOps
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
IMG_SIZE=128
folder_path= Path("./Data/Face/Aiden")
image_paths= sorted(folder_path.glob("image_*.jpg"))
print("이미지 개수",len(image_paths))
print(image_paths[:10])
```

```python
data=[]

for path in image_paths:
    img=Image.open(path)

    img=img.convert("L")

    img.thumbnail((IMG_SIZE,IMG_SIZE))

    canvas = Image.new("L",(128,128),0)

    x = (IMG_SIZE - img.size[0]) // 2
    y = (IMG_SIZE - img.size[1]) // 2

    canvas.paste(img, (x, y))
    img_array=np.array(canvas)

    data.append(img_array)

data=np.array(data)

print(data.shape)
```

```python
plt.figure(figsize=(10, 4))

for i in range(len(data)):
    plt.subplot(3, 6, i + 1)
    plt.imshow(data[i], cmap="gray")
    plt.axis("off")
    plt.title(f"img {i}")

plt.show()
```

```python
data= data/255.0
```

```python
data= data.reshape(-1,128,128,1)

print(data.shape)
```

```python
IMG_SIZE=128
folder_path2= Path("./Data/Face/Andrew")
image_paths2= sorted(folder_path2.glob("image_*.jpg"))
print("이미지 개수",len(image_paths2))
print(image_paths2[:10])
```

```python
from PIL import Image
import numpy as np

IMG_SIZE = 128

andrew_data = []

for path in image_paths2:
    # 1. 이미지 열기 + 흑백 변환
    img = Image.open(path).convert("L")

    # 2. 원본 비율 유지하면서 128x128 안에 들어가게 축소
    img.thumbnail((IMG_SIZE, IMG_SIZE))

    # 3. 검은색 128x128 캔버스 생성
    canvas = Image.new("L", (IMG_SIZE, IMG_SIZE), 0)

    # 4. 중앙에 붙일 위치 계산
    x = (IMG_SIZE - img.size[0]) // 2
    y = (IMG_SIZE - img.size[1]) // 2

    # 5. 검은 배경 중앙에 이미지 붙이기
    canvas.paste(img, (x, y))

    # 6. numpy 배열 변환
    img_array = np.array(canvas)

    # 7. 리스트에 추가
    andrew_data.append(img_array)

andrew_data = np.array(andrew_data)

print(andrew_data.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

for i in range(len(andrew_data)):
    plt.subplot(3, 6, i + 1)
    plt.imshow(andrew_data[i], cmap="gray")
    plt.axis("off")
    plt.title(f"Andrew {i}")

plt.show()
```

```python


andrew_data= andrew_data/255.0

andrew_data= andrew_data.reshape(-1,128,128,1)

print(andrew_data.shape)
```

```python
andrew_data.shape
```

```python
from tensorflow import keras
from tensorflow.keras import layers
```

```python
from sklearn.model_selection import train_test_split

train_input, test_input = train_test_split(
    andrew_data,
    test_size=0.2,
    random_state=42
)

print(train_input.shape)
print(test_input.shape)
```

```python
from tensorflow import keras
from tensorflow.keras import layers

autoencoder = keras.Sequential()
```

```python
autoencoder.add(
    layers.Conv2D(
        32,
        kernel_size=3,
        activation='relu',
        padding='same',
        input_shape=(128, 128, 1)
    )
)

autoencoder.add(
    layers.MaxPooling2D(
        pool_size=2,
        padding='same'
    )
)

autoencoder.add(
    layers.Conv2D(
        64,
        kernel_size=3,
        activation='relu',
        padding='same'
    )
)

autoencoder.add(
    layers.MaxPooling2D(
        pool_size=2,
        padding='same'
    )
)

autoencoder.add(
    layers.Conv2D(
        128,
        kernel_size=3,
        activation='relu',
        padding='same'
    )
)

autoencoder.add(
    layers.MaxPooling2D(
        pool_size=2,
        padding='same'
    )
)
```

```python
autoencoder.add(
    layers.Conv2D(
        128,
        kernel_size=3,
        activation='relu',
        padding='same'
    )
)

autoencoder.add(
    layers.UpSampling2D(2)
)

autoencoder.add(
    layers.Conv2D(
        64,
        kernel_size=3,
        activation='relu',
        padding='same'
    )
)

autoencoder.add(
    layers.UpSampling2D(2)
)

autoencoder.add(
    layers.Conv2D(
        32,
        kernel_size=3,
        activation='relu',
        padding='same'
    )
)

autoencoder.add(
    layers.UpSampling2D(2)
)
```

```python
autoencoder.add(
    layers.Conv2D(
        1,
        kernel_size=3,
        activation='sigmoid',
        padding='same'
    )
)
```

```python
autoencoder.compile(
    optimizer='adam',
    loss='mse'
)
```

```python
history = autoencoder.fit(
    train_input,
    train_input,
    epochs=20,
    batch_size=4,
    validation_data=(test_input, test_input)
)
```
