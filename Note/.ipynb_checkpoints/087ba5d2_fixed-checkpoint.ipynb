### 선형회귀를 위한 clss만들기

```python
class LRModel:
    #Property
    
    #constructor
    def __init__(self):
            
        self.w=1.0
        self.b=1.0
    #method
    #정방향 계산
    def forpass(self,x):
        y_hat=x*self.w + self.b 
        return y_hat

    
    def backprop(self,x,err):
        w_grad=x*err
        b_grad=1*err
        return w_grad,b_grad

    #훈련
    def fit(self,x,y,epoch):
        for _ in range(epoch):
            for x_i,y_i in zip(x,y):
                y_hat = self.forpass(x_i)
                err=y_i-y_hat
                w_grad,b_grad = self.backprop(x_i,err)
                self.w += w_grad
                self.b +=b_grad

                return self.w,self.b

    
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 5, 7, 9, 11]

lrmodel=LRModel()
w,b=lrmodel.fit(x,y,100)
print(w,b)
```

#### 딥 러닝 모델 만들기

```python
from tensorflow import keras
from tensorflow.keras.layers import Input
```

```python
model = keras.Sequential()
model.add(Input(shape=(784,))) # 입력층
model.add(keras.layers.Dense(100,activation ='relu'))# 은닉층은 여러개가
model.add(keras.layers.Dense(10,activation='softmax'))#출력층

#손실함수
model.compile(
    optimizer= 'adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
# 데이터 훈련
model.fit(
    np.array(train_data),
    np.array(train_label),
    epochs = 10
)
```

## 출력층
- 은닉층에서 전달된 데이터를 최종적으로 처리하여 결과를 내는 층입니다.
- 분류 문제라면 `softmax`나 `sigmoid` 같은 활성화 함수가 사용되어 예측 확률을 계산합니다.

## 손실함수
- 출력층에서 나온 결과와 실제 값을 비교하여 오류를 계산합니다.
- 손실함수는 예측이 얼마나 잘못되었는지를 평가합니다.

```python
from tensorflow import keras

(train_input, train_target),(test_input, test_target) = \
    keras.datasets.fashion_mnist.load_data()
```

```python
from sklearn.model_selection import train_test_split

# 셀을 여러 번 실행해도 train_target 길이가 꼬이지 않도록 원본 데이터를 다시 불러옵니다.
(train_input, train_target), (test_input, test_target) = keras.datasets.fashion_mnist.load_data()

# 이미지 픽셀값 정규화 + 28x28 이미지를 784개 feature로 펼치기
train_scaled = (train_input / 255.0).reshape(-1, 28 * 28)
test_scaled = (test_input / 255.0).reshape(-1, 28 * 28)

# train 데이터를 실제 학습용 / 검증용으로 분리
sub_input, val_scaled, sub_target, val_target = train_test_split(
    train_scaled,
    train_target,
    test_size=0.2,
    random_state=42
)

print("sub_input:", sub_input.shape)
print("val_scaled:", val_scaled.shape)
print("sub_target:", sub_target.shape)
print("val_target:", val_target.shape)
```

```python
# layer 만들기
# Input 레이어를 따로 두면 input_shape 경고가 나오지 않습니다.
dense1 = keras.layers.Dense(100, activation='sigmoid')
dense2 = keras.layers.Dense(10, activation='softmax')

model = keras.Sequential([
    keras.layers.Input(shape=(784,)),
    dense1,
    dense2
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
# 모델 학습
history = model.fit(
    sub_input,
    sub_target,
    epochs=10,
    validation_data=(val_scaled, val_target)
)
```

```python
# 테스트 데이터로 최종 평가
model.evaluate(test_scaled, test_target)
```
