```python
from tensorflow import keras
from sklearn.model_selection import train_test_split

(train_input, train_target),(test_input,test_target)=\
    keras.datasets.fashion_mnist.load_data()

train_scaled = train_input.reshape(-1,28,28,1) # (Dimension, Row, Coloumn, Channel)

train_scaled,val_scaled,trian_target,val_target = \
    train_test_split(
        train_scaled,
        train_target,
        test_size=0.2,
        random_state=42
    )
```

#### 합성곱층 만들기

```python
model = keras.Sequential()
```

```python
model.add(
    keras.layers.Cov2D(
        32,
        kernel_size = 3, # 보통 정사각형으로 구성
        activation = 'relu', #활성화 함수
        padding = 'same',
        input_shape=(28,28,1)
        
    )
)
```

#### 풀링층

```python
model.add(
    keras.layers.MaxPooling2D(2)
)
```

### 두번째 합성곱층

```python
model.add(
    keras.layers.Cov2D(
        64,
        kernel_size = 3, # 보통 정사각형으로 구성
        activation = 'relu', #활성화 함수
        padding = 'same',
        
    )
)
```

```python
model.add(
    keras.layers.MaxPooling2D(2)
)
```

#### Dense층

```python
#입력층
model.add(
    keras.layers.Flatten()
)
#은닉층
model.add(
    keras.layers.Dense(
        100,
        activation= 'relu'
    )
)

# Drop out 층
model.add(
    keras.layers.Dropout(0.5)
)

# 출력층
```

```python
model.compile(
    loss= = 'sparse_categorical_crossentropy',
optimizer='adam',
    metrics=['accuracy']
)

checkpoint_cb=    keras.callbacks.ModelCheckpoint("..Data/best_cnn_model.csv")
```
